#### Missingness

```python
# Check if any remaining columns have more than half missing values
missing_counts = df.isnull().sum()
missing_counts = missing_counts[missing_counts > len(df) / 2]
missing_counts.sort_values(ascending=False)
```

```python
# Chose to keep the license variable and transform it to binary
# A non-listed license in itself can be a useful feature in predicting nightly price
# The presence of a license can be a useful feature in predicting nightly price, as it may indicate that the host is more professional and has invested in their property.
df["has_license"] = df["license"].notna().astype(int)
df = df.drop(columns="license")
```

#### Impute missing categorical variables

```python
# Impute missing values in categorical variables with 'missing'
cat_cols = df.select_dtypes(include=["object", "string"]).columns
df[cat_cols] = df[cat_cols].fillna("missing")

# Check to ensure no categorical variables have missing values
df[cat_cols].isnull().sum()
```

#### Low Variability

```python
from sklearn.feature_selection import VarianceThreshold

# Drop the target variable before checking variance
df_num = df.select_dtypes(include=['number']).drop(columns="price_quote_price_per_night")

# Set your variance threshold (e.g., 0.01 is a common starting point)
threshold = 0.01
selector = VarianceThreshold(threshold=threshold)
selector.fit(df_num)

# Get boolean mask of features that pass the threshold
mask = selector.get_support()

# Get list of all column names
all_features = df_num.columns

# Features that were removed (low variance)
low_variability_features = all_features[~mask]

# Show the names of low-variability columns
print("Low variability features (variance < {:.2f}):".format(threshold))
print(low_variability_features.tolist())

# Latitude and Longitude are low variance, but they are important features for predicting nightly price, so they will be kept.
# These degrees of variance look small, but one degree of latitude is approximately 69 miles, and one degree of longitude is approximately 55.8 miles in Nashville, TN.
```

```python
# Check for low variability using a near-zero variance test instead of VarianceThreshold
# VarianceThreshold compares raw variances, which depend on units. It flagged latitude and longitude
# (variance ~0.002) only because they are measured in degrees, even though they span ~45 km and are ~70% unique.
# A near-zero variance test is unit-independent: a variable is flagged only if one value dominates
# (freq_ratio > 19, i.e. 95/5 split) AND it has few distinct values (pct_unique < 10%).
# Result: no variables were flagged, so none are removed for low variability.
num = df.select_dtypes(include=["number"]).drop(columns="price_quote_price_per_night")

nzv = pd.DataFrame({
    "freq_ratio": num.apply(lambda s: s.value_counts().iloc[0] / s.value_counts().iloc[1]),
    "pct_unique": num.nunique() / len(num) * 100
})
nzv["near_zero_var"] = (nzv["freq_ratio"] > 19) & (nzv["pct_unique"] < 10)
nzv.sort_values("freq_ratio", ascending=False)
```

```python
# Check for low variability in categorical variables
# A categorical variable has low variability if one category makes up more than 95% of rows
flagged = False
for col in cat_cols:
    top_freq = df[col].value_counts(normalize=True).iloc[0]
    if top_freq > 0.95:
        print(f"{col} ({top_freq:.1%})")
        flagged = True
if not flagged:
    print("No categorical variables with low variability")
```

#### Filter Monthly Rentals

```python
# Remove monthly rentals (minimum stay of 28+ nights)
# Nashville defines a short-term rental as a stay of fewer than 30 nights, so listings requiring a month or more
# are a different market with different pricing. 28 is used as the cutoff to also catch 4-week minimum stays.
# Listings with a missing minimum_nights are kept (NaN >= 28 is False) and will be imputed after the train/test split.
print("Rows before filter:", len(df))
df = df[~(df["minimum_nights"] >= 28)]
print("Rows after filter:", len(df))
```
